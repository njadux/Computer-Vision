```python
# Saving image path
path = '/content/drive/MyDrive/Rana Jadu.png'
```

```python
# load image
img = Image.open(path)
```

# Extracting data from img

```python
# Extracting text in string format
extraced = pytesseract.image_to_string(img)
```

```python
print(extraced)
```

```python
# Remove leading and trailing whitespaces from a string
extraced_text = extraced.strip()
```

# Comparison

```python
# Get user input
entered_data = input("Enter your signature: ")
```

```python
# Comparing the user input to the predefined data
if extraced_text == entered_data:
  print("Access")
else:
  print("denied")
```
